# Inverse-Square Law for DDE with saddle node bifurcation
The inverse-square law for DDEs with saddle node bifurcation states that there is an inverse-square relationship between the overshoot of the control parameter $q$ from the bifurcation value $q^b$ , and the duration, $t_e$, of overshoot. 
The relationship is stated as
\begin{equation}
a^2(\tau)\kappa(\tau)(q^{\text{max}}-q^b)t_e^2\le 4, \,\,q^{\mathrm{max}},q^b,a(\tau),\kappa(\tau)\in\mathbb{R},t_{e}\in\mathbb{R^{+}_{0}},\tau>0.
\end{equation}
$q^{\max}$ is the maximum value that $q$ attains during the overshoot. The delay dependent terms $a^2(\tau)$, and $\kappa(\tau)$ result from the centre manifold reduction of $\frac{dy}{dt}=f(y(t),y_{\tau}(t),q), \tau>0,t\in\mathbb{R},y(t)\in\mathbb{R}^n,q\in\mathbb{R}$, where notationally $y_{\tau}(t)=y(t-\tau)$. The use of $y_{\tau}$ to mean the same thing as $y_{\tau}(t)$ will also be adopted. The term $y_{\tau}$ doesn't mean the deriviative of $y$ with respeect to $\tau$.  It is to be noted that $f$ is sufficiently differentiable.  For the case $n=1$, where the state variable $y(t)\in\mathbb{R}^{n=1}$,
\begin{align}\tag{1}
a(\tau)&=\frac{f_{q}(y^b,y^b,q^b)}{(1+\tau f_{y_{\tau}}(y^b,y^b,q^b))},\,\,\,(1+\tau f_{y_{\tau}}(y^b,y^b,q^b))\ne 0,\\
a(\tau)\kappa(\tau)&= \frac{(f_{yy}(y^b,y^b,q^b)+2f_{yy_{\tau}}(y^b,y^b,q^b)+f_{y_\tau y_\tau}(y^b,y^b,q^b))}{2(1+\tau f_{y_{\tau}}(y^b,y^b,q^b))},\,\,\,(1+\tau f_{y_{\tau}}(y^b,y^b,q^b))\ne 0.
\end{align}

The notation $f_{y}$ means we take the partial derivative of the function $f(y,y_{\tau},q)$ with respect to $y$ variable. Likewise, $f_{yy}$ means the partial derivative of the function $f(y,y_{\tau},q)$ with respect to $y$ twice. Moreover, the various variable(s) that are in the subscript of $f$, as in the terms above, naturally mean the partial derivative of $f$ with respect to those variable(s) in the text direction. 

Substituing the terms $a(\tau)$, and $a(\tau)\kappa(\tau)$ in the inverse-square law yields
\begin{equation}
\left[\frac{f_{q}(y^b,y^b,q^b)\left(f_{yy}(y^b,y^b,q^b)+2f_{yy_{\tau}}(y^b,y^b,q^b)+f_{y_\tau y_\tau}(y^b,y^b,q^b)\right)}{2(1+\tau f_{y_{\tau}}(y^b,y^b,q^b))^2}\right](q^{\mathrm{max}}-q^b)t_{e}^2\le 4,\,\,\,(1+\tau f_{y_{\tau}}(y^b,y^b,q^b))\ne 0.
\end{equation}

Equivalently,[I want to label the equation below but cant do it why??]
\begin{align}
\left[\frac{f_{q}\left(f_{yy}+2f_{yy_{\tau}}+f_{y_\tau y_\tau}\right)}{2(1+\tau f_{y_{\tau}})^2}\right](q^{\mathrm{max}}-q^b)t_{e}^2&\le 4,\,\,\,(1+\tau f_{y_{\tau}})\ne 0,\tag{*}\\
f_{q}\left(f_{yy}+2f_{yy_{\tau}}+f_{y_\tau y_\tau}\right)(q^{\mathrm{max}}-q^b)&\le \left[\frac{8(1+\tau f_{y_{\tau}})^2}{t_{e}^2}\right],\,\,\,(1+\tau f_{y_{\tau}})\ne 0,\tag{**}
\end{align}
where the functions above are evaluated at the bifurcation point, therefore, for convenience we have suppressed the argument.

# Plotting the inverse-square law relationship for DDEs
In this notebook we will be plotting the inverse square law relationship for 1 dimensional DDEs.
In particular functions of the form,
$\begin{equation}y'=\pm q\pm (y_\tau-c_1)(y-c_2)\end{equation}, y(t)\in\mathbb{R},q\in\mathbb{R},\tau>0,c_1,c_2\in\mathbb{R}$,
will be considered. Notationally $y'=\frac{dy}{dt}$.

We will refer to the following functions by their corresponding cases:
- $\begin{equation}y'=q+(y_{\tau}-c_1)(y-c_2)\end{equation}$ ; Case 1
- $\begin{equation}y'= -q+(y_{\tau}-c_1)(y-c_2)\end{equation}$ ; Case 2
- $\begin{equation}y'=-q -(y_{\tau}-c_1)(y-c_2)\end{equation}$ ; Case 3
- $\begin{equation}y'=q-(y_{\tau}-c_1)(y-c_2)\end{equation}$ ; Case 4


# Case 1
Let us consider the case $\begin{equation}y'=q+(y{(t-\tau)}-c_1)(y-c_2)\end{equation}$.

Since the delay does not affect the position of the bifurcation point, one can be compute the bifurcation point by solving for the value of $\bar{y}=y^b$ in 
\begin{equation}
q+(\bar{y}-c_1)(\bar{y}-c_2)=0,
\end{equation}
such that for some $q=q^b$ the equation has only one solution. 

Expanding the terms out in $\bar{y}$ and using the quadratic fomula one can see
\begin{align}
\bar{y}^2-(c_1+c_2)\bar{y}+c_1c_2+q&=0\\
\implies\bar{y}&=\frac{(c_1+c_2)\pm\sqrt{(c_1-c_2)^2-4q}}{2}.
\end{align}
The equation above will have only one solution when $(c_1-c_2)^2-4q=0$, zero solutions when $(c_1-c_2)^2-4q<0$, and two solutions when $(c_1-c_2)^2-4q>0$. Therefore, the bifurcation value $q^b$ is given by $(c_1-c_2)^2-4q^b=0\iff q^b=\left(\frac{c_1-c_2}{2}\right)^2$, in turn $y^b=\left(\frac{c_1+c_2}{2}\right)$ is the bifurcation value that occurs when $q=q^b$. Hence, 
\begin{align}
q^b&=\left(\frac{c_1-c_2}{2}\right)^2,\\
y^b&=\left(\frac{c_1+c_2}{2}\right).
\end{align}


For the function $f(y,y_{\tau},q)=q+(y_{\tau}-c_1)(y-c_2)$ at the equilibrium point $(y^b,y^b,q^b)$,
\begin{align}
f_{q}=1,f_{y_{\tau}}=\frac{c_1-c_2}{2},f_{yy}=f_{y_{\tau}y_{\tau}}=0,f_{y_{\tau}y}=1.
\end{align}

Therefore, terms $a(\tau)$, and $a(\tau)\kappa(\tau)$ using (1) evaluate to
\begin{align}
a(\tau)&=\frac{1}{\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2},\,\,\,1+\tau\left(\frac{c_1-c_2}{2}\right)\ne 0,\\
a(\tau)\kappa(\tau)&=\frac{2}{2\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2},\,\,\,1+\tau\left(\frac{c_1-c_2}{2}\right)\ne 0.\\
\end{align}

Subsequently, the inverse-square realtionship (**) is 
\begin{align}
\frac{2}{2\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2}\left(q^{\mathrm{max}}-\left(\frac{c_1-c_2}{2}\right)^2\right)t_{e}^2&\le 4,\\
q^{\mathrm{max}}&\le\frac{4}{t_{e}^2}\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2+\left(\frac{c_1-c_2}{2}\right)^2
\end{align}

In the python script below we explore the relationship between $q^{\mathrm{max}}$, and $t_e$ for varying values of $\tau\in \{1,2,\cdots,10\}$, along with chosen $c_1$, and $c_2$ values.

Note if $c_1=c_2$ then the $f_{y_{\tau}}=0$, which implies that the delay has no bearing in the inverse-square law. Therefore, we assume that $c_1\ne c_2$.

## Plot for Case 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Fixed Parameter Value
c1 = -1
c2 = 2
```

```python
# Values of tau in [1,2,...,10]
list_tau = np.arange(1, 11) 
```

```python
# defining d just to make computation easier
# t_e is the duration which is on the x axis
d=(c1-c2)/2
t_e = np.linspace(0.001, 10, 1000)
```

```python
# Plotting the inverse sqaure relationship

plt.figure(figsize=(8, 5)) # Initialise the plot
for tau in list_tau:
    q_max = (
        4 * (1 + tau * d)**2 / t_e**2
        + d**2)
    line, = plt.plot(t_e, q_max, label=rf'$\tau={tau}$') #Returns a 2D object so the 'line,' syntax only takes the 1D line
    plt.fill_between(t_e, q_max, 0, color=line.get_color(), alpha=0.15) #Shade the region between the curve(s) and the baseline y=0.
plt.xlabel(r'$t_e$')
plt.ylabel(r'$q^{\max}$')
plt.xlim(0, 10)
plt.ylim(0, 20)
plt.legend()
plt.grid()
plt.show()

# note the flatline is the result of the denominator of a(tau), and a(tau)kappa(tau) being zero (i.e 1+tau*f_{y_{tau}}=0)
```

# Case 2

Consider the case $\begin{equation}y'=-q+(y_{\tau}-c_1)(y-c_2)\end{equation}$.

The bifurcation point can be computed by solving for the value of $\bar{y}=y^b$ in 
\begin{equation}
-q+(\bar{y}-c_1)(\bar{y}-c_2)=0,
\end{equation}
such that for some $q=q^b$ the equation has only one solution. 

Expanding the terms out in $\bar{y}$ and using the quadratic fomula one can see
\begin{align}
\bar{y}^2-(c_1+c_2)\bar{y}+c_1c_2-q&=0\\
\implies\bar{y}&=\frac{(c_1+c_2)\pm\sqrt{(c_1-c_2)^2+4q}}{2}.
\end{align}
The equation above will have only one solution when $(c_1-c_2)^2+4q=0$, zero solutions when $(c_1-c_2)^2+4q<0$, and two solutions when $(c_1-c_2)^2+4q>0$. Therefore, the bifurcation value $q^b$ is given by $(c_1-c_2)^2+4q^b=0\iff q^b=-\left(\frac{c_1-c_2}{2}\right)^2$, in turn $y^b=\left(\frac{c_1+c_2}{2}\right)$ is the bifurcation value that occurs when $q=q^b$. Hence, 
\begin{align}
q^b&=-\left(\frac{c_1-c_2}{2}\right)^2,\\
y^b&=\left(\frac{c_1+c_2}{2}\right).
\end{align}


For the function $f(y,y_{\tau},q)=-q+(y_{\tau}-c_1)(y-c_2)$ at the equilibrium point $(y^b,y^b,q^b)$,
\begin{align}
f_{q}=-1,f_{y_{\tau}}=\frac{c_1-c_2}{2},f_{yy}=f_{y_{\tau}y_{\tau}}=0,f_{y_{\tau}y}=1.
\end{align}

Therefore, terms $a(\tau)$, and $a(\tau)\kappa(\tau)$ using (1) evaluate to
\begin{align}
a(\tau)&=\frac{-1}{\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2},\,\,\,1+\tau\left(\frac{c_1-c_2}{2}\right)\ne 0,\\
a(\tau)\kappa(\tau)&=\frac{2}{2\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2},\,\,\,1+\tau\left(\frac{c_1-c_2}{2}\right)\ne 0.\\
\end{align}

Subsequently, the inverse-square realtionship (**) is 
\begin{align}
\frac{-2}{2\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2}\left(q^{\mathrm{max}}+\left(\frac{c_1-c_2}{2}\right)^2\right)t_{e}^2&\le 4,\\
q^{\mathrm{max}}&\ge-\frac{4}{t_{e}^2}\left(1+\tau\left(\frac{c_1-c_2}{2}\right)\right)^2-\left(\frac{c_1-c_2}{2}\right)^2
\end{align}

In the python script below we explore the relationship between $q^{\mathrm{max}}$, and $t_e$ for varying values of $\tau\in \{1,2,\cdots,10\}$, along with chosen $c_1$, and $c_2$ values.

Note if $c_1=c_2$ then the $f_{y_{\tau}}=0$, which implies that the delay has no bearing in the inverse-square law. Therefore, we assume that $c_1\ne c_2$.

## Plot for Case 2


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Computing the constants 

c1, c2 = 0, 2
d = (c1 - c2) / 2
```

```python
# Initialising the array for the x-axis
t_e = np.linspace(0.1, 5, 500)
```

```python
#Plot of the graph

for tau in range(1, 11):
    boundary = -(4 / t_e**2) * (1 + tau * d)**2 - d**2
    plt.plot(t_e, boundary, label=f"tau={tau}")
    plt.fill_between(t_e, boundary, 5, alpha=0.15)

plt.xlabel("t_e")
plt.ylabel("q_max")
plt.xlim(0,5)
plt.ylim(-60, 5)
plt.legend()
plt.savefig("simple_plot2.png", dpi=150)

```

## Case 3

Consider the case $\begin{equation}y'=-q-(y_{\tau}-c_1)(y-c_2)\end{equation}$.

The bifurcation point can be computed by solving for the value of $\bar{y}=y^b$ in 
\begin{equation}
-q-(\bar{y}-c_1)(\bar{y}-c_2)=0,
\end{equation}
such that for some $q=q^b$ the equation has only one solution. 

Expanding the terms out in $\bar{y}$ and using the quadratic fomula one can see
\begin{align}
-(\bar{y}^2-(c_1+c_2)\bar{y}+c_1c_2+q)&=0\\
\implies\bar{y}&=\frac{(c_1+c_2)\pm\sqrt{(c_1-c_2)^2-4q}}{2}.
\end{align}
The equation above will have only one solution when $(c_1-c_2)^2-4q=0$, zero solutions when $(c_1-c_2)^2-4q<0$, and two solutions when $(c_1-c_2)^2-4q>0$. Therefore, the bifurcation value $q^b$ is given by $(c_1-c_2)^2-4q^b=0\iff q^b=\left(\frac{c_1-c_2}{2}\right)^2$, in turn $y^b=\left(\frac{c_1+c_2}{2}\right)$ is the bifurcation value that occurs when $q=q^b$. Hence, 
\begin{align}
q^b&=\left(\frac{c_1-c_2}{2}\right)^2,\\
y^b&=\left(\frac{c_1+c_2}{2}\right).
\end{align}


For the function $f(y,y_{\tau},q)=-q-(y_{\tau}-c_1)(y-c_2)$ at the equilibrium point $(y^b,y^b,q^b)$,
\begin{align}
f_{q}=-1,f_{y_{\tau}}=-\left(\frac{c_1-c_2}{2}\right),f_{yy}=f_{y_{\tau}y_{\tau}}=0,f_{y_{\tau}y}=-1.
\end{align}

Therefore, terms $a(\tau)$, and $a(\tau)\kappa(\tau)$ using (1) evaluate to
\begin{align}
a(\tau)&=\frac{-1}{\left(1-\tau\left(\frac{c_1-c_2}{2}\right)\right)^2},\,\,\,1-\tau\left(\frac{c_1-c_2}{2}\right)\ne 0,\\
a(\tau)\kappa(\tau)&=\frac{-2}{2\left(1-\tau\left(\frac{c_1-c_2}{2}\right)\right)^2},\,\,\,1-\tau\left(\frac{c_1-c_2}{2}\right)\ne 0.\\
\end{align}

Subsequently, the inverse-square realtionship (**) is 
\begin{align}
\frac{2}{2\left(1-\tau\left(\frac{c_1-c_2}{2}\right)\right)^2}\left(q^{\mathrm{max}}-\left(\frac{c_1-c_2}{2}\right)^2\right)t_{e}^2&\le 4,\\
q^{\mathrm{max}}&\le\frac{4}{t_{e}^2}\left(1-\tau\left(\frac{c_1-c_2}{2}\right)\right)^2+\left(\frac{c_1-c_2}{2}\right)^2
\end{align}

In the python script below we explore the relationship between $q^{\mathrm{max}}$, and $t_e$ for varying values of $\tau\in \{1,2,\cdots,10\}$, along with chosen $c_1$, and $c_2$ values.

Note if $c_1=c_2$ then the $f_{y_{\tau}}=0$, which implies that the delay has no bearing in the inverse-square law. Therefore, we assume that $c_1\ne c_2$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
c1 = 0
c2 = 2
list_tau = np.arange(1, 11)
```

```python
d=(c1-c2)/2
t_e = np.linspace(0.001, 10, 1000)
```

```python

plt.figure(figsize=(8, 5))
for tau in list_tau:
    q_max = (
        4 * (1 - tau * d)**2 / t_e**2
        + d**2)
    line, = plt.plot(t_e, q_max, label=rf'$\tau={tau}$') #Returns a 2D object so the 'line,' syntax only takes the 1D line
    plt.fill_between(t_e, 0, q_max, color=line.get_color(), alpha=0.15) #Shade the region below the line defined above.

plt.xlabel(r'$t_e$')
plt.ylabel(r'$q^{\max}$')
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.legend()
plt.grid()
plt.show()
```
